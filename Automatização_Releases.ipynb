## Verificação de serviços mais solicitados e porcentagens desses serviços

```python
tp_servicos = df_relatorio['Serviço'] #Definindo uma variável para o campo de Serviços
tp_servicos.unique()#Mostrando todos os tipos de serviços
```

```python
tp_servicos.unique().shape[0]#Mostra a quantidade total de serviços
```

```python
# Conta a frequência de cada serviço no campo Serviço
import matplotlib.pyplot as plt
contagem_servicos = df_relatorio['Serviço'].value_counts()

# Seleciona os top 10 serviços mais solicitados
top_10_servicos = contagem_servicos.head(10)

# Definir uma lista de cores para as barras
cores = ['#FF5733', '#33FF57', '#3357FF', '#F333FF', '#33FFF5', '#F5FF33', '#FF33A1', '#33FFA1', '#A133FF', '#FF9133']

# Plotando o gráfico de colunas horizontais com cores personalizadas
plt.figure(figsize=(10, 6))
ax = top_10_servicos.sort_values().plot(kind='barh', color=cores)
plt.title('Top 10 Serviços mais Solicitados')
plt.xlabel('Número de Solicitações')
plt.ylabel('Serviço')

# Adicionar os valores das contagens
for i in ax.patches:
    ax.text(i.get_width() + 0.5, i.get_y() + i.get_height() / 2, str(int(i.get_width())), ha='left', va='center')

plt.show()

print("Top 10 Serviços mais Solicitados")
print(top_10_servicos)
```

```python
#Calcula a porcentagem de cada um dos serviços do top 10
top_10_servicos_porcentagem = (top_10_servicos / total_solicitacoes) * 100

#Formata para 2 casas decimais
top_10_servicos_porcentagem = top_10_servicos_porcentagem.round(2)

#Soma a conclusão das porcentagens dos top 10 serviços
soma_porcentagem_top_10 = top_10_servicos_porcentagem.sum()

print("Top 10 servicos mais solicitados com porcentagem")
print(top_10_servicos_porcentagem)
print(f"Soma das porcentagens dos top 10 serviços: {soma_porcentagem_top_10:.2f}%")
```

## Verificação dos Órgãos mais solicitados e porcentagem de conclusão de serviços em cada Órgão

```python
tp_orgao = df_relatorio['Órgão'] #Definindo uma variável para o campo Órgão
tp_orgao.unique() #Mostra todos os tipos de Órgão
```

```python
# Conta a frequência de cada órgão no campo Órgão
orgaos_contagem = df_relatorio['Órgão'].value_counts()

# Seleciona os top 5 órgãos mais frequentes
top_5_orgaos = orgaos_contagem.head(5)

# Calcula as porcentagens de cada um dos órgãos
porcentagem_orgaos = (top_5_orgaos / total_solicitacoes) * 100

# Formata para 2 casas decimais
porcentagem_orgaos = porcentagem_orgaos.round(2)

# Soma as porcentagens dos top 5 órgãos
soma_porcentagem_orgaos = porcentagem_orgaos.sum()

# Adiciona a porcentagem dos demais órgãos
porcentagem_outros = 100 - soma_porcentagem_orgaos

print("Top 5 Órgãos mais frequentes com porcentagens:")
print(porcentagem_orgaos)
print(f"Soma das porcentagens dos top 5 Órgãos: {soma_porcentagem_orgaos:.2f}%")

# Lista de cores para os gráficos
cores = ['#1B4C8C', '#143340', '#618C78', '#E1B964', '#D29146', '#CD412D']

# Gráfico de porcentagem de solicitações concluídas por órgão
fig, ax = plt.subplots(figsize=(14, 2))  # Ajuste o tamanho da figura

# Dados para o gráfico
orgaos = list(porcentagem_orgaos.index) + ['Outros']
porcentagens = list(porcentagem_orgaos.values) + [porcentagem_outros]

# Adiciona a barra horizontal acumulada
ax.barh([0], [porcentagens[0]], height=0.4, color=cores[0], edgecolor='black')  # Ajuste a altura da barra
for i in range(1, len(porcentagens)):
    ax.barh([0], [porcentagens[i]], height=0.4, left=sum(porcentagens[:i]), color=cores[i], edgecolor='black')

# Adiciona texto com a porcentagem no gráfico
for i, (percent, orgao) in enumerate(zip(porcentagens, orgaos)):
    ax.text(sum(porcentagens[:i]) + percent / 2, 0, f"{percent:.1f}%", ha='center', va='center', color='white' if percent > 5 else 'black')

# Configurações do gráfico
ax.set_xlim(0, 100)
ax.set_title('Solicitações Recebidas - Por Órgão')
ax.get_yaxis().set_visible(False)
ax.get_xaxis().set_visible(False)

# Legenda
handles = [plt.Rectangle((0,0),1,1, color=cores[i]) for i in range(len(cores))]
ax.legend(handles, orgaos, loc='upper center', bbox_to_anchor=(0.5, -0.1), ncol=3)

plt.show()

plt.show()
```

```python
#Filtrar o dataframe para incluir apenas solicitações dos top 5 órgãos
df_top_5_orgaos = df_relatorio[df_relatorio['Órgão'].isin(top_5_orgaos.index)]

#Filtrar dataframe para incluir apenas as solicitações concluídas dentro dos top 5 órgãos
df_concluidas_top_5 = df_top_5_orgaos[filtro_concluidas]

#Contar a quantidade de solicitações concluídas e total dentro dos top 5 órgãos
qtd_concluidas_top_5 = len(df_concluidas_top_5)
total_top_5_orgaos = len(df_top_5_orgaos)

#Calcular a porcentagem de solicitações concluidas dentro dos top 5 órgãos
porcentagem_orgaos_top_5 = (qtd_concluidas_top_5 / total_top_5_orgaos) * 100

print(f"Quantidade de solicitações concluídas dentro dos top 5 órgãos: {qtd_concluidas_top_5}")
print(f"Porcentagem de solicitações concluídas dentro dos top 5 órgãos: {porcentagem_orgaos_top_5:.2f}%")
```

```python
# Filtrar o dataframe para incluir apenas solicitações concluídas
df_concluidas = df_relatorio[df_relatorio['Status da solicitação'] == 'FINALIZADA']
total_concluidas = len(df_concluidas)

# Filtrar o dataframe para incluir apenas solicitações dos top 5 órgãos
df_top_5_orgaos = df_concluidas[df_concluidas['Órgão'].isin(top_5_orgaos.index)]

# Contar a quantidade de solicitações concluídas por órgão dentro dos top 5
qtd_concluidas_top_5 = df_top_5_orgaos['Órgão'].value_counts()

# Calcular a porcentagem de solicitações concluídas por órgão, em relação ao total geral de solicitações concluídas
porcentagem_orgaos_top_5 = (qtd_concluidas_top_5 / total_concluidas) * 100

# Filtrar o dataframe para incluir solicitações dos órgãos não no top 5
df_outros_orgaos = df_concluidas[~df_concluidas['Órgão'].isin(top_5_orgaos.index)]

# Calcular a porcentagem de solicitações concluídas nos órgãos não incluídos no top 5
qtd_concluidas_outros = len(df_outros_orgaos)
porcentagem_outros = (qtd_concluidas_outros / total_concluidas) * 100

# Adicionar a porcentagem dos órgãos restantes como "Outros"
porcentagem_orgaos_top_5['Outros'] = porcentagem_outros

# Imprimir quantidades e porcentagens
print(f"Quantidade de solicitações concluídas dentro dos top 5 órgãos:\n{qtd_concluidas_top_5}")
print(f"Porcentagem de solicitações concluídas por órgão:\n{porcentagem_orgaos_top_5}")

# Dados para o gráfico
orgaos = porcentagem_orgaos_top_5.index
porcentagens = porcentagem_orgaos_top_5.values

# Número de cores necessárias
num_cores = len(orgaos)

# Lista de cores personalizadas para os gráficos
cores = ['#1B4C8C', '#143340', '#618C78', '#E1B964', '#D29146', '#CD412D']

# Gráfico de porcentagem de solicitações concluídas por órgão
fig, ax = plt.subplots(figsize=(14, 2))  # Ajuste o tamanho da figura

# Adiciona a barra horizontal acumulada
ax.barh([0], [porcentagens[0]], height=0.4, color=cores[0], edgecolor='black')  # Ajuste a altura da barra
for i in range(1, len(porcentagens)):
    ax.barh([0], [porcentagens[i]], height=0.4, left=sum(porcentagens[:i]), color=cores[i], edgecolor='black')

# Adiciona texto com a porcentagem no gráfico
for i, (percent, orgao) in enumerate(zip(porcentagens, orgaos)):
    ax.text(sum(porcentagens[:i]) + percent / 2, 0, f"{percent:.1f}%", ha='center', va='center', color='white' if percent > 5 else 'black')

# Configurações do gráfico
ax.set_xlim(0, 100)
ax.set_title('Porcentagem de Solicitações Concluídas por Órgão (Top 5 e Outros)')
ax.get_yaxis().set_visible(False)
ax.get_xaxis().set_visible(True)

# Legenda
handles = [plt.Rectangle((0,0),1,1, color=cores[i]) for i in range(num_cores)]
ax.legend(handles, orgaos, loc='upper center', bbox_to_anchor=(0.5, -0.15), ncol=3)

plt.show()
```

## Verificação de usos de cada Canal

```python
tp_canais = df_relatorio['Canal']
tp_canais.unique()
```

```python
# Contar a quantidade de solicitações recebidas por canal
solicitacoes_por_canal = df_relatorio['Canal'].value_counts()

# Contar a quantidade de solicitações concluídas em cada Canal
solicitacoes_concluidas_por_canal = df_relatorio[df_relatorio['Status da solicitação'].isin(['FINALIZADA'])]['Canal'].value_counts()

# Criar um DataFrame para armazenar os resultados
resultados_canais = pd.DataFrame({
    'Total Solicitações': solicitacoes_por_canal,
    'Solicitações Concluídas': solicitacoes_concluidas_por_canal
}).fillna(0).astype(int)

# Calcular a porcentagem de solicitações finalizadas em cada canal
resultados_canais['Porcentagem Finalizadas'] = (resultados_canais['Solicitações Concluídas'] / resultados_canais['Total Solicitações']) * 100

# Calcular a porcentagem de solicitações recebidas por canal
resultados_canais['Porcentagem Recebidas'] = (resultados_canais['Total Solicitações'] / len(df_relatorio)) * 100

# Formatar a porcentagem para 2 casas decimais
resultados_canais = resultados_canais.round(2)

# Print dos resultados
print("Resultados por Canal de Atendimento")
print(resultados_canais)

# Preparar os dados para o gráfico
porcentagem_recebidas = resultados_canais['Porcentagem Recebidas']
labels = resultados_canais.index

# Criar o gráfico de pizza
fig, ax = plt.subplots(figsize=(8, 8))
wedges, texts, autotexts = ax.pie(porcentagem_recebidas, labels=labels, autopct='%1.1f%%', colors=['#19374B', '#1F67A6', '#6EA087', '#FFD273'], startangle=140)

# Melhorar a legibilidade do gráfico
plt.setp(autotexts, size=10, weight="bold", color='white')
plt.setp(texts, size=12, weight="bold")

ax.set_title('Distribuição de Solicitações Recebidas por Canal')
plt.show()
```

## Criação do Excel do código com todas as informações

```python
# Instale o xlsxwriter, se ainda não estiver instalado
!pip install xlsxwriter

# Dados para o Excel
dados = {
    'Resumo Geral': {
        'Total de Solicitações': [df_relatorio.shape[0]],
        'Quantidade de Solicitações Concluídas': [qtd_concluidas],
        'Porcentagem de Solicitações Concluídas': [f"{porcentagem_concluidas:.2f}%"]
    },
    'Top 10 Serviços': {
        'Serviço': top_10_servicos.index,
        'Quantidade': top_10_servicos.values
    },
    'Top 10 Serviços %': {
        'Serviço': top_10_servicos.index,
        'Quantidade': top_10_servicos.values,
        'Porcentagem': top_10_servicos_porcentagem.values
    },
    'Top 5 Órgãos %': {
        'Órgão': porcentagem_orgaos.index,
        'Porcentagem': porcentagem_orgaos.values
    },
    'Concluídas nos Top 5 Órgãos': {
        'Órgão': [f"Quantidade: {qtd_concluidas_top_5}", f"Porcentagem: {porcentagem_orgaos_top_5:.2f}%"]
    },
    'Resultados por Canal': resultados_canais.reset_index()
}

# Criar um escritor de Excel
with pd.ExcelWriter(nome_arquivo, engine='xlsxwriter') as writer:
    # Adicionar cada DataFrame como uma planilha
    for sheet_name, data in dados.items():
        df = pd.DataFrame(data)
        df.to_excel(writer, sheet_name=sheet_name, index=False)

```
